```python
from pymongo import MongoClient
from transformers import AutoTokenizer, AutoModel
import torch
import numpy as np

# 1) Conectar a la base
db = mongo_client["spare_parts_db"]
collection = db["repuestos"]

# 2) Modelo HF para generar embedding de la consulta
modelo = "sentence-transformers/all-MiniLM-L6-v2"
tokenizer = AutoTokenizer.from_pretrained(modelo)
model = AutoModel.from_pretrained(modelo)

def embed_text(text):
    tokens = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    with torch.no_grad():
        output = model(**tokens)
    emb = output.last_hidden_state.mean(dim=1).squeeze().numpy().tolist()
    return emb

# 3) Generar embedding de la consulta del usuario
query_vector = embed_text("pastillas de freno para Fiat Uno")

# 4) Consulta vectorial en Mongo:
resultados = collection.aggregate([
    {
        "$vectorSearch": {
            "queryVector": query_vector,
            "vector": "embedding",   # nombre del campo vectorial
            "numCandidates": 100,
            "limit": 5
        }
    }
])

# 5) Imprimir resultados
for r in resultados:
    print(r)
```

## Ingestion (chunks) and RAG

```python
!pip install pypdf langchain_community
from langchain_community.document_loaders import PyPDFLoader

# We consider a large PDF file
pdf_path = "Understanding_Climate_Change.pdf"

loader = PyPDFLoader(pdf_path)
pdf_documents = loader.load() # Each document corresponds actually to a page
print(len(pdf_documents), "loaded")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

def replace_t_with_space(list_of_documents):
    for doc in list_of_documents:
        doc.page_content = doc.page_content.replace('\t', ' ')  # Replace tabs with spaces
    return list_of_documents


# Split documents into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, chunk_overlap=100)

texts = text_splitter.split_documents(pdf_documents)
cleaned_texts = replace_t_with_space(texts)
print(len(cleaned_texts), "chunks")
```

```python
uuids = [str(uuid4()) for _ in range(len(cleaned_texts))]
for id, doc in zip(uuids, cleaned_texts):
    doc._id = id
    doc.id = id
```

```python
# We use a vector store for the chunks
# chroma_vectorstore = Chroma.from_documents(cleaned_texts, OpenAIEmbeddings())
# my_retriever = chroma_vectorstore.as_retriever(search_kwargs={"k": 5})

mongo_vectorstore =  MongoDBAtlasVectorSearch.from_documents(
        cleaned_texts,
        collection=MONGODB_COLLECTION,
        embedding=embeddings,
        index_name="otro_indice",
        relevance_score_fn="cosine",
    )

mongo_vectorstore.create_vector_search_index(dimensions=384)

#my_retriever = mongo_vectorstore.as_retriever(search_kwargs={"k": 5})

my_retriever = mongo_vectorstore.as_retriever(search_kwargs={"k": 5, "search_type": "mmr"})
```

```python
# Helper function for printing docs
def pretty_print_docs(docs):
    print(
        f"\n{'-' * 100}\n".join(
            [f"Document {i + 1}:\n\n" + d.page_content for i, d in enumerate(docs)]
        )
    )

test_query = "What is the main cause of climate change?"
context_docs =  my_retriever.invoke(test_query)
pretty_print_docs(context_docs)
```
